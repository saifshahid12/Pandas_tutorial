Data frame through dict

```python
import pandas as pd
d = {"a":[1,2,3,4,5],"b":[1,2,3,4,5],"d":[1,2,3,4,5]}
var1 = pd.DataFrame(d)
print(var1)
```

Columns function in dataframe :
and change index number :

```python
import pandas as pd
d = {"a":[1,2,3,4,5],"b":[1,2,3,4,5],"d":[1,2,3,4,5]}
var1 = pd.DataFrame(d,columns=["b"], index=["s","g","h","f","e"])     # for get single column w use key word here 
print(var1)
```
