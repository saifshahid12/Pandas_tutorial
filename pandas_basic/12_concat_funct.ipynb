Concat function :
with Series :



```python
# concat contain list and give out put in series form when we merge
import pandas as pd 
sr1 = pd.Series([1,2,3,4])
sr2 = pd.Series([11,21,31,41])

pd.concat([sr1,sr2])
```

cancate with series or daataframe :


```python
# using axis=

import pandas as pd 
sr3 = pd.Series([1,2,3,4])
sr4 = pd.Series([11,21,31,41])
sr5 = pd.Series([1,2,3,4])
sr6 = pd.Series([11,21,31,41])


pd.concat([sr3,sr4,sr5,sr6],axis=1) #axis work on separate rows and collumns  show in terminal
```
