dropna and fillna

```python
import pandas as pd

var =  pd.read_csv("E:\\pandas_basic\\text1.csv")
var
```

Now drop NAN value from existing file:

```python
# remove NAN value from file using .dropna func
var.dropna()
```

drop missing colomns : column contain value =1:

```python
# remove NAN columnns from files using .dropna
var.dropna(axis=1)
```

drop missing row from file : row contain valiue 0:

```python
# drop missing NAN rows from files using dropna . and axis will be 0 for row always 
var.dropna(axis=0)
```

Remove any single NAN line from file:

```python
# remove just NAN block which is NAN :using dropna(how=)
var.dropna(how="any")       # using any
```

remove single row which consist all NAN : using dropna(how="all")

```python
# remove 1 single line which have no element in the file 
var.dropna(how="all") #remove line number 29 from
```

remove just NAN block from columns:using dropna(subset="")

```python
# drop column CITY nan values : using d .dropna(subset="City")
var.dropna(subset="City")
```

remove all NAN values and create new table  :using .dropna(inplace=True)

```python
# remove all NAN item from list :using .dropna(inplace=True)
var.dropna(inplace=True)
var
```

thresh function :
work on single nan value


```python
var.dropna(thresh=1)
```

```python
var1 =  pd.read_csv("E:\\pandas_basic\\text1.csv")
var1
```

use filnna funct to write NAN place in table 

```python
# where was nan , now it show python 
var1.fillna("python")
```

give value to nan place any single column or all :through fillna 


```python
# now we change every column and row where nan is available there we write this :
# 
var1.fillna({"ID":"saif","Name":"python","Gender":"monkey","Department":"hazara","City":"Mansehra",	"Marks":"10,00000"})
```

Forward filling method :

```python
# see the line number 29 / here lune number 28 forward in 29
var1.fillna(method="ffill")      # forward fill
```

Backward filling
:


```python
# backward fillna method use to copy front to back 
# we can se the line number 29 where line number 30 is copied in line 29
var1.fillna(method="bfill")    # backkwardfill
```
