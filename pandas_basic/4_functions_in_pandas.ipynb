Insert function :

```python
import pandas as pd
var = pd.DataFrame({"A":[1,2,3,4,5],"B":[9,8,7,6,5]})
# var
var.insert(1,"python",var["A"])    # insert through duplicate values of "A"
var
```

```python
var.insert(1,"python_1",[20,30,40,50,60])
var                              # insert a new list in out put
```

Slicing throgh index number :

```python
var = pd.DataFrame({"C":[1,2,3,4,5],"D":[6,7,8,9,1]})
var["python_12"]=var["D"][:3]    # find limited values through index slicing .
var
```

```python
var = pd.DataFrame({"F":[1,2,3,4,5],"G":[6,7,8,9,0],"C":[20,30,40,50,60]})
var
```

```python
var1= var.pop("C")   # delete colum through key value
var
```
