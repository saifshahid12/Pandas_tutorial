Melt function:


```python
# create table 
# horizantel
import pandas as pd 
var = pd.DataFrame({"days":[1,2,3,4,5],"eng":[10,12,13,14,15],"maths":[20,30,40,50,60]})
var
```

make table in verticle: through melt()

```python
# conver into verticle :
pd.melt(var)
```

create ID at top of table :

```python
# create id at the top of tbale (days)
pd.melt(var,id_vars=["days"]) #create new id name of days 
```

change name of existing ID "value"

```python
# create new name saif at existing name (value):
var1 = pd.DataFrame({"days":[1,2,3,4,5],"eng":[10,12,13,14,15],"maths":[20,30,40,50,60]})
pd.melt(var1,value_name="saif")
```
