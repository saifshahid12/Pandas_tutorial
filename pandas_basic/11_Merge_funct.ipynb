Merge function :use on= function 


```python
# In merge there is same values ocuur then the out put will be show just 1 time in same column
# for example A is two time :

import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4,5],"B":[11,12,13,14,15]})
var2 = pd.DataFrame({"A":[1,2,3,4,5],"C":[21,22,23,24,25]})
pd.merge(var1,var2,on="A")
```

```python
# fuctions in merge :

# how="inner"
# how="outer",indicator=True
# how="left"
# how="right"
# pd.merge(var3,var4,left_index=True , right_index = True) it show all value when key are same 
# pd.merge(var3,var4,left_index=True , right_index = True)suffixes="name","python") it change the columns heading name  
import pandas as pd
var3 = pd.DataFrame({"A":[1,2,3,4,6],"B":[11,12,13,14,15]})     #thus is the left data var3.
var4 = pd.DataFrame({"A":[1,2,3,4,5],"C":[21,22,23,24,25]})     #thus is the right data var4.

# pd.merge(var3,var4,how="outer",indicator=True)
pd.merge(var3,var4,left_index=True , right_index = True,suffixes=("name","python"))
```
