Series :

```python
import pandas as pd

x = [3,5,6,7,8,9,8,]
var = pd.Series(x)
print(var)
print(type(var))
print(var[2])
```

changing index in alphabets and change into float
AND use name :


```python
x = [3,5,6,7,8,9,8,]
var = pd.Series(x,index=['a','b','c','d','e','f','j'],dtype=float,name="python")
print(var)
```

Dictionary:

```python
dic={"name":['python','jawa','c++','c'],"prop":[12,13,14,15],"Rank":[1,2,3,4]}
var1 = pd.Series(dic)
print(var1)
```

Single Series :


```python
a = pd.Series(12)
print(a)
print(type(a))
```

using index number :


```python
a = pd.Series(12,index=[1,2,3,4,5,6,7])
print(a)
```

Indexing series in miltiple dimenson:


```python
s1 = pd.Series(12,index=[1,2,3,4,5,6,7])
s2 = pd.Series(12,index=[1,2,3,4])
print(s1 +s2)
# extra index will be NAN in pandas 
```
