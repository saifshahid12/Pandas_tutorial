```python
import pandas as pd

var = pd.read_csv("E:\\pandas_basic\\text1.csv")
var
```

USE replace func:


```python
  # line no 46 mohsin raza replace with python 
  
var.replace(to_replace="Mohsin Raza",value="python") 
```

Replace numbers using .replace:


```python
# replace numbers using .replace .
# 1 to 9 replace with "hi"

var.replace([1,2,3,4,5,6,7,8,9],"hi")
```

Replace through dictionary : using .replace and regex=True


```python
# replace column "NAME" where {A-Z} all are reolace with hello 
# [A-Z] mean that all alphabets in it

var.replace({"Name":"[A-Z]"},"hello",regex=True)
```

forward replace method :


```python
# male will be replaced by the value before it , like female 
var.replace("Male",method="ffill")
```
