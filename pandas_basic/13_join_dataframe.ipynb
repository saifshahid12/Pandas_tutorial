join dataframs:


```python
import pandas as pd
# join separate data frames 
# using .join
var1 = pd.DataFrame({"A":[1,2,3,4],"B":[10,20,30,40]})
var2 = pd.DataFrame({"C":[11,12,13,14],"D":[50,60,70,80]})
var1.join(var2)    
```

Change index numbers :through index=

```python
# change index from 1,2,3,4 to a,b,c,d 
# we haveto change index in both line if we change index of var3 the output will be NAN for colm c and d

var3 = pd.DataFrame({"A":[1,2,3,4],"B":[10,20,30,40]},index=["a","b","c","d"])
var4 = pd.DataFrame({"C":[11,12,13,14],"D":[50,60,70,80]},index=["a","b","c","d"])
var3.join(var4)    
```

change sides through how=

```python
var7 = pd.DataFrame({"A":[1,2,3,4],"B":[10,20,30,40]})
var8 = pd.DataFrame({"C":[11,14],"D":[50,60,]})
var7.join(var8,how="right")  
# functions :
# how="left"  
# how="right"  
# how="outer"  for union
# how="inner"  for intersection 
# suffix=   for remove key error
```
